# 주피터 노트북에서 깃허브 사용하기 for Mac

## 깃허브 로그인하기
> GitHub 로그인 -> 상단 (+버튼) 클릭 -> New repository -> Repository name 작성 -> public으로 설정

## 맥북 터미널 접근
> command + space을 입력 후 Terminal 입력

> cd ~/Desktop/.... 과 같이 현재 작업중인 폴더 접근
> ex) cd ~/Desktop/Data_analysis/project

> pwd를 입력하여 현재 위치 확인

## Git 접속하기
### 1. git init  # 이하 해당 폴더를 Git으로 관리하는 폴더로 지정
> exe) initialized empty Git repository....


### 2. git status  # 현재 파일의 상태를 나타냄 ex) modified:... -> 파일이 수정됨을 나타냄
> exe) ~~.ipynb


### 3. git add .  # 폴더 안에 있는 파일들을 commit을 위한 상태로 바꿈(.은 현재 폴더를 뜻함)


### 4. git commit -m "설명"  # add를 통해 commit으로 올라온 파일을 Git에 업로드할 준비를 함(-m은 ""을 통해 저장할 내용을 적음)
> exe) create mode....


### 5. GitHub에 새로 만든 repository에 들어가면 git remote add origin https.....으로 나온 부분 복사/붙여넣기  


### 6. git branch -M main  # 프로젝트의 기둥을 -M 옵션으로 이름을 main으로 바꿈


### 7. git push -u origin main  # 작업 공간에서 commit한 기록을 GitHub에 업로드함
> 간혹 로그인을 요구하는 경우가 있음. 그럴 경우 다음과 같이 따라하면 됨
> 1. 5번에서 복사했던 링크에서 "https://github.com/아이디.." 부분 중 아이디를 확인
> 2. 아이디를 요구하는 경우 위에서 확인한 아이디 작성
> 3. 비밀번호를 요구할 경우 아래를 계속 진행
> 4. 프로필 -> setting -> Developer settings -> personal access tokens -> Tokens(classic) -> 토큰 생성 후 토큰을 복사하여 비밀번호 부분에 복사

## 수정 후 GitHub Update 방법

#### 1. 파일 수정 후 저장

#### 2. 터미널을 통해 Git에 접속할 폴더 접속 
ex) cd ~/Desktop...

#### 3. git add .

#### 4. git commit -m "수정한 내용"

#### 5. git push
